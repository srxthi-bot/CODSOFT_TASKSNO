Step 1: Import libraries

```python
print("Descriptive Statistics:")
df.describe()
```

Step 9 — Find Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Missing value percentage:")
print(missing_percentage.round(2))
```

Step 10 — Find Duplicate Records

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate records:", duplicate_count)
```

```python
print("Duplicate records:")
display(df[df.duplicated()])
```

Step 11 — Check Inconsistent Entries

```python
categorical_columns = df.select_dtypes(include="object").columns
print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
for column in categorical_columns:
    print(f"\nUnique values in {column}:")
    print(df[column].unique())
```

Step 12 — Create a quick data-quality report

```python
data_quality = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique()
})

display(data_quality)
```

Step 13 — Handle missing values

```python
df_clean = df.copy()

print("Cleaning process started...")
```

13.1 — Handle missing

```python
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())
```

13.2 — Handle missing embarked

```python
df_clean["embarked"] = df_clean["embarked"].fillna(
    df_clean["embarked"].mode()[0]
)
```

13.4 — Handle deck

```python
df_clean["deck"] = df_clean["deck"].astype("object")
df_clean["deck"] = df_clean["deck"].fillna("Unknown")
```

Step 14 — Check missing values again

```python
print("Missing values after handling nulls:")
print(df_clean.isnull().sum())
```

Step 15 — Remove duplicate records

```python
duplicates_before = df_clean.duplicated().sum()
print("Duplicates before removal:", duplicates_before)
df_clean = df_clean.drop_duplicates()
duplicates_after = df_clean.duplicated().sum()
print("Duplicates after removal:", duplicates_after)
```

Step 16 — Correct data types

```python
print("Data types before correction:")
print(df_clean.dtypes)
```

```python
categorical_columns = [
    "class",
    "who",
    "embarked",
    "deck",
    "embark_town",
    "alive"
]

for column in categorical_columns:
    df_clean[column] = df_clean[column].astype("category")
```

```python
print("Data types after correction:")
print(df_clean.dtypes)
```

Step 17 — Final data-quality check

Missing values

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

Duplicates

```python
print("\nDuplicate records after cleaning:")
print(df_clean.duplicated().sum())
```

Dataset shape

```python
print("\nCleaned dataset shape:")
print(df_clean.shape)
```

First few rows

```python
print("\nCleaned dataset:")
display(df_clean.head())
```

Step -18 : Save the Cleaned Dataset

```python
df_clean.to_csv("cleaned_titanic.csv", index=False)
print("Cleaned dataset saved successfully!")
```
